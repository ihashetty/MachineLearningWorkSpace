```python
df.fillna(df["salary"].mean(), inplace=True)
print(df)
```

```python
df['bonus']=5000
print(df)
```

```python

df.loc[df['department']=='HR', 'salary'] *= 1.1
df.loc[df['department']=='IT', 'salary'] *= 1.2
df.loc[df['department']=='Finance', 'salary'] *= 1.15
print(df)
```

```python
df['department'] = df['department'].replace('Operations',5)
print(df)
df.to_csv('data1.csv', index=False)
```

```python
df1=pd.read_csv('dept.csv')
print(df1)

df=df.merge(df1, on='department', how='left')
print(df)

df2=pd.merge(df, df1, left_on='department', right_on='department')
print(df2)
```

```python
import matplotlib.pyplot as plt
plt.bar(df['experience'], df['salary'])
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(df['salary'], bins=10, color='blue', alpha=1)
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Salary Distribution')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.plot(df['salary'], df['age'],color='green', marker='o', linestyle='None', alpha=0.7)
plt.xlabel('Salary')
plt.ylabel('Age')
plt.title('Salary vs Age')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['salary'], df['age'], color='green', alpha=0.7)
plt.xlabel('Salary')
plt.ylabel('Age')
plt.title('Salary vs Age')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.pie(
    df['salary'],
    labels=df['age'],
    autopct=lambda pct: f'{int(round(pct/100 * df["salary"].sum(), 0))}'
)
plt.title('Salary vs Age')
plt.show()
```
